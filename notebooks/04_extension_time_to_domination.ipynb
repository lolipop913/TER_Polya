# Temps de stabilisation de la domination

Les notebooks précédents ont étudié l’existence et la probabilité empirique de la domination permanente dans une urne de Pólya généralisée.

Nous cherchons maintenant à comprendre la dynamique temporelle de cette domination.

Le théorème de Janson affirme qu’il existe une probabilité strictement positive que les boules noires restent dominantes pour tout temps futur. Cependant, il ne décrit pas directement la manière dont cette domination se stabilise au cours du temps.

L’objectif de ce notebook est donc d’étudier expérimentalement le temps nécessaire pour que la domination noire devienne stable sur un horizon fini.

## Définition empirique du temps de stabilisation

L’événement théorique de domination permanente est :

$$
A = \{B_n > W_n,\ \forall n \geq 0\}.
$$

Comme les simulations sont réalisées sur un horizon fini \(T\), nous introduisons une version empirique du temps de stabilisation.

On définit :

$$
\tau_T = \min \{n : B_k > W_k,\ \forall k \in [n,T]\}.
$$

Autrement dit, \(\tau_T\) est le premier instant à partir duquel les boules noires restent dominantes jusqu’à la fin de la simulation.

Si une trajectoire ne devient jamais durablement dominante sur l’horizon étudié, on considère que \(\tau_T\) n’est pas défini pour cette trajectoire.

```python
import sys
import os

sys.path.append(os.path.abspath(".."))

import numpy as np
import matplotlib.pyplot as plt

from src.urn import PolyaUrn
```

## Première estimation sur une trajectoire

Nous commençons par calculer le temps de stabilisation sur une seule trajectoire.

Les paramètres utilisés satisfont les hypothèses du théorème de Janson :

$$
B_0 > W_0
\quad \text{et} \quad
m_b > m_w.
$$

```python
urn = PolyaUrn(
    balls={"black": 10, "white": 5},
    reinforcements={"black": 2, "white": 1}
)

urn.simulate(1000)

tau = urn.stabilization_time()

print(f"Empirical stabilization time: {tau}")
```

## Distribution empirique des temps de stabilisation

Une seule trajectoire ne permet pas de conclure, car le processus reste aléatoire.

Nous répétons donc l’expérience sur un grand nombre de trajectoires indépendantes afin d’observer la distribution empirique des temps de stabilisation.

```python
n_simulations = 1000
n_steps = 1000

taus = []

for _ in range(n_simulations):

    urn = PolyaUrn(
        balls={"black": 10, "white": 5},
        reinforcements={"black": 2, "white": 1}
    )

    urn.simulate(n_steps)

    tau = urn.stabilization_time()

    if tau is not None:
        taus.append(tau)

taus = np.array(taus)

print(f"Number of stabilized trajectories: {len(taus)}")
print(f"Total simulations: {n_simulations}")
print(f"Mean stabilization time: {taus.mean():.2f}")
print(f"Median stabilization time: {np.median(taus):.2f}")
```

## Interprétation

L’histogramme décrit la distribution empirique du temps à partir duquel la domination noire devient stable jusqu’à l’horizon final \(T\).

Lorsque les temps de stabilisation sont faibles, cela signifie que la domination devient rapidement durable.

Lorsque les temps sont plus élevés, cela indique que les fluctuations initiales continuent d’influencer la trajectoire pendant une période plus longue.

Cette analyse prolonge le théorème de Janson : le théorème affirme l’existence d’une probabilité positive de domination permanente, tandis que cette expérience étudie la vitesse empirique à laquelle cette domination se stabilise.
